```python
import requests
from bs4 import BeautifulSoup
import wikipediaapi
```

```python
def get_plot_from_wikipedia(url, output_file):
    """
    Fetch the plot of a novel from a given Wikipedia page URL and write it to a text file.

    Parameters:
    - url: The URL of the Wikipedia page.
    - output_file: The path to the output text file.
    """
    # Initialize Wikipedia API
    wiki = wikipediaapi.Wikipedia('MyProjectName (merlin@example.com)', 'en')
    page_name = url.split("/wiki/")[-1]
    page = wiki.page(page_name)

    # If page is valid, try to get plot summary from 'Plot' section
    plot_text = ""
    if page.exists():
        sections = page.sections
        for section in sections:
            if section.title.lower() == "plot":
                plot_text = section.text
                break
            
    # Write the plot summary to the output file
    with open(output_file, 'w', encoding='utf-8') as file:
        file.write(plot_text)
```

```python
# Example usage
url = "https://en.wikipedia.org/wiki/Dune_Messiah"
output_file = "./dune_plot_summary.txt"
get_plot_from_wikipedia(url, output_file)
print(f"Plot summary written to {output_file}")
```
