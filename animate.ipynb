```python
import cv2
import numpy as np
from pathlib import Path 
```

```python
image_1 = Path("/home/joffreyma/Projets/verbose-octo-carnival/images_to_anim/1.webp")
image_2 = Path("/home/joffreyma/Projets/verbose-octo-carnival/images_to_anim/2.webp")
```

```python
img = cv2.imread(image_1)
img2 = cv2.imread(image_2)
```

```python
height, width, layers = img.shape
height, width, layers
```

```python
# Define video writer for exporting to MP4
height, width, layers = img.shape
fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # Codec
video = cv2.VideoWriter('./zoom_effect.mp4', fourcc, 30.0, (width, height))
```

```python
# Generate zoom frames
for i in range(24*5):  # 50 frames for the zoom animation
    zoom_factor = 1 + i * 0.001  # Gradual zoom effect
    zoomed_img = cv2.resize(img, None, fx=zoom_factor, fy=zoom_factor)
    
    # Crop the zoomed image to the original size
    zoom_height, zoom_width, _ = zoomed_img.shape
    start_x = (zoom_width - width) // 2
    start_y = (zoom_height - height) // 2
    cropped_img = zoomed_img[start_y:start_y + height, start_x:start_x + width]
    
    video.write(cropped_img)  # Write the frame to the video
```

```python
# Apply Gaussian blur using OpenCV
blurred_image = cv2.GaussianBlur(cropped_img, (51, 51), 10)  # (15, 15) is the kernel size
for i in range(24*3):
    video.write(blurred_image)
```

```python
# Release the video writer
video.release()
```

```python
# Generate panning frames
frames = 24*10
for i in range(frames):  # 50 frames for the panning animation
    shift = int(i * 0.5 * (width / frames))  # Gradually move to the right
    pan_img = np.roll(img2, shift, axis=1)  # Pan image horizontally
    video.write(pan_img)  # Write the frame to the video
```

```python
# Release the video writer
video.release()
```
